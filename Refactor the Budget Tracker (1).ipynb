```python
# Store monthly expenses in a list of dictionaries

expenses = [
    {"category": "Food", "item": "Rice", "amount": 10500},
    {"category": "Food", "item": "Beans", "amount": 7000},
    {"category": "Food", "item": "Indomie", "amount": 5000},
    {"category": "Transport", "item": "T-fare", "amount": 4000},
    {"category": "Transport", "item": "Fuel", "amount": 2500},
    {"category": "Airtime", "item": "Phone credit", "amount": 4500},
    {"category": "Rent", "item": "House rent", "amount": 30000},
    {"category": "Food", "item": "Eggs", "amount": 2000},
    {"category": "Subscription", "item": "Light", "amount": 10000},

    # Records with invalid or missing amounts
    {"category": "Food", "item": "Invalid expense", "amount": "n/a"},
    {"category": "Food", "item": "Unknown expense", "amount": None},
    {"category": "Food", "item": "Missing amount"}
]

     # Set the monthly budget for each category
budgets = {
    "Food": 20000,
    "Transport": 8000,
    "Airtime": 5000,
    "Rent": 30000,
    "Subscription": 15000
}
```

```python
     # Check whether an amount is a valid number

def is_valid_amount(amount):
    try:
        float(amount)
        return amount is not None
    except (ValueError, TypeError):
        return False
```

```python
# Calculate spending totals for each category

def total_by_category(expenses):
    totals = {}

    for expense in expenses:

        # Skip records without a valid amount
        amount = expense.get("amount")

        if not is_valid_amount(amount):
            continue

        category = expense["category"]
        amount = float(amount)

        # Add the amount to the category total
        if category in totals:
            totals[category] += amount
        else:
            totals[category] = amount

    return totals
```

```python
# Compare spending with budgets and return a summary

def summarise(totals, budgets):
    summary = []

    for category, budget in budgets.items():
        spent = totals.get(category, 0)

        if spent > budget:
            difference = spent - budget
            status = f"OVER by ₦{difference:,.0f}"
        else:
            status = "OK"

        # Create one summary line for each category
        line = (
            f"{category} spent ₦{spent:,.0f} "
            f"of ₦{budget:,.0f} - {status}"
        )

        summary.append(line)

    return summary
```

```python
# Return expenses above a chosen threshold

def filter_expenses(expenses, threshold):
    return [
        expense for expense in expenses
        if is_valid_amount(expense.get("amount"))
        and float(expense["amount"]) > threshold
    ]
```

```python
# Calculate category totals
totals = total_by_category(expenses)

# Generate the budget summary
report = summarise(totals, budgets)

print("MONTHLY BUDGET SUMMARY")
print("----------------------")

for line in report:
    print(line)

# Filter expenses above ₦5,000
large_expenses = filter_expenses(expenses, 5000)

print("\nEXPENSES ABOVE ₦5,000")
print("---------------------")

for expense in large_expenses:
    print(f"{expense['item']}: ₦{float(expense['amount']):,.0f}")
```

```python
# Test that category totals are correct
assert totals["Food"] == 24500
assert totals["Transport"] == 6500
assert totals["Airtime"] == 4500

# Test that invalid amounts are handled
assert is_valid_amount("n/a") == False
assert is_valid_amount(None) == False
assert is_valid_amount(5000) == True

# Test that the summary contains the overspending status
assert any("OVER by ₦4,500" in line for line in report)

# Test the threshold filter
filtered = filter_expenses(expenses, 10000)

assert all(float(e["amount"]) > 10000 for e in filtered)

print("All tests passed!")
```
